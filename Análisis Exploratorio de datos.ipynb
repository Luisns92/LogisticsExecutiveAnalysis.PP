### Formateo de datos

#### Creación de funciones para formatear datos

```python
def convert_to_lowercase(df):
    """
    Convierte a minúsculas todas las columnas de texto
    de un DataFrame.
    """
    df = df.copy()

    text_cols = df.select_dtypes(include=["object", "string"]).columns

    for col in text_cols:
        df[col] = df[col].str.lower()

    return df
```

```python
def convert_dates(df, date_columns):
    """
    Convierte columnas de fecha a tipo datetime.
    """
    df = df.copy()

    for col in date_columns:
        df[col] = pd.to_datetime(df[col])

    return df
```

```python
def convert_float(df, int_columns):
    """
    Convierte columnas de tipo int a tipo float con 2 decimales (para revenues, currencies).
    """
    df = df.copy()

    for col in int_columns:
        df[col] = df[col].astype('float').round(2)

    return df
```

#### Aplicación de funciones y cambios

```python
# Establecer todos los datasets en minúsculas
df_customers = convert_to_lowercase(df_customers)
df_drivers = convert_to_lowercase(df_drivers)
df_trucks = convert_to_lowercase(df_trucks)
df_trailers = convert_to_lowercase(df_trailers)
df_routes = convert_to_lowercase(df_routes)
df_loads = convert_to_lowercase(df_loads)
df_trips = convert_to_lowercase(df_trips)
df_fuel_purchases = convert_to_lowercase(df_fuel_purchases)
df_maintenance_records = convert_to_lowercase(df_maintenance_records)
df_delivery_events = convert_to_lowercase(df_delivery_events)
```

```python
# Establecer las fechas en todos los datasets
df_customers = convert_dates(df_customers, ['contract_start_date'])
df_drivers = convert_dates(df_drivers, ['hire_date', 'date_of_birth'])
df_trucks = convert_dates(df_trucks, ['acquisition_date'])
df_trailers = convert_dates(df_trailers, ['acquisition_date'])
df_loads = convert_dates(df_loads, ['load_date'])
df_trips = convert_dates(df_trips, ['dispatch_date'])
df_fuel_purchases = convert_dates(df_fuel_purchases, ['purchase_date'])
df_maintenance_records = convert_dates(
    df_maintenance_records, ['maintenance_date'])
df_delivery_events = convert_dates(df_delivery_events, [
                                   'scheduled_datetime', 'actual_datetime'])
```

```python
# Cambiar tipo de dato 'object' por 'float64' para revenues
df_customers['annual_revenue_potential'] = df_customers['annual_revenue_potential'].astype(
    'float64').round(2)
```

```python
# Cambiar boleano por numero binario
df_delivery_events["on_time_flag"] = df_delivery_events['on_time_flag'].astype(
    int)
```
